```sql
-- =====================================================
-- PROFIT BY YEAR
-- =====================================================

SELECT
    YEAR(TO_DATE(order_date,'d/M/yyyy')) AS year,
    ROUND(SUM(profit),2) AS total_profit
FROM enriched_orders
GROUP BY YEAR(TO_DATE(order_date,'d/M/yyyy'))
ORDER BY year



-- =====================================================
-- PROFIT BY YEAR + PRODUCT CATEGORY
-- =====================================================
SELECT
    YEAR(TO_DATE(order_date,'d/M/yyyy')) AS year,
    category,
    ROUND(SUM(profit),2) AS total_profit
FROM enriched_orders where category is not null
GROUP BY
    YEAR(TO_DATE(order_date,'d/M/yyyy')),
    category
ORDER BY year

-- =====================================================
-- PROFIT BY CUSTOMER
-- =====================================================

SELECT
    customer_name,
    ROUND(SUM(profit),2) AS total_profit
FROM enriched_orders
GROUP BY customer_name
ORDER BY total_profit DESC;



-- =====================================================
-- PROFIT BY CUSTOMER + YEAR
-- =====================================================

SELECT
    customer_name,
    YEAR(TO_DATE(order_date,'d/M/yyyy')) AS year,
    ROUND(SUM(profit),2) AS total_profit
FROM enriched_orders where customer_name is NOT NULL
GROUP BY
    customer_name,
    YEAR(TO_DATE(order_date,'d/M/yyyy'))
    ORDER BY 3,2 desc
```
